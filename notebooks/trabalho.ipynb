```python
import pandas as pd

df = pd.read_csv('../dados/dataset_trabalho.csv')

display(df)
```

```python
df.isnull().sum()
```

```python
# Criar uma cópia para preservar o dataset original
df_tratado = df.copy()

# Tratar a idade com a mediana
mediana = df_tratado['idade'].median()
df_tratado['idade'] = df_tratado['idade'].fillna(mediana)

# Tratar a cidade com "Desconhecido"
df_tratado['cidade'] = df_tratado['cidade'].fillna('Desconhecido')

# Exibir o resultado
display(df_tratado)
```

```python
df_tratado = df_tratado.drop(columns=['observacoes'])

display(df_tratado)
```
