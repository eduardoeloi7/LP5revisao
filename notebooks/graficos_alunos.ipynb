```python
# importando cada biblioteca
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('../dados/dados_alunos.csv')
df
```

```python
df['idade'].plot.hist(
    bins=10 , 
    title='Histograma de idades', 
    figsize=(8, 4)
    )
# plt.title('')
plt.xlabel('idades')
plt.ylabel('Frequencia')
plt.show()
```

```python
# agrupando dados

# df['genero'].value_counts()
df['curso'].value_counts()
```

```python
df['genero'].value_counts().plot.bar(
    figsize=(8, 4)
)
plt.xticks(rotation=0)
plt.title('Quantidade por gênero')
plt.xlabel('genero')
plt.ylabel('quantidade')
plt.show()
```

```python
df['genero'].value_counts().plot.barh(
    figsize=(8, 4)
)
plt.xticks(rotation=0)
plt.title('Quantidade por gênero')
plt.xlabel('quantidade')
plt.ylabel('genero')
plt.show()
```

```python
df['genero'].value_counts().plot.pie(
    colors={"#f4a5fc", "#75c6f5"},
    figsize=(8,8),
    autopct='%1.1f%%',
    wedgeprops=dict(width=0.9)
    
)
plt.title('Quantidade por gênero')

plt.show()
```
