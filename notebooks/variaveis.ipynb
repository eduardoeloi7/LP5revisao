# tipos de variaveis 

## numericas

- discretas: assume numeros inteiros 
- continua: assume numeros com intervalos de valores(float)(ex:altura e peso)


```python
print("Idade (variavel numerica DISCRETA)") 
print(df['Idade'])
print("Altura (variavel numerica CONTINUA)") 
print(df["Altura"]) 
```

# variaveis categoricas

- Nominais: Não tem ordem ou hierarquia (ex: curso)
- Ordinais: Possuem ordem natural, ordem lógica ou hierarquica. (ex: escolaridade)

```python
print("curso (variavel categorica nominal)") 
print(df['Curso'])
print("escolaridade (variavel categorica ordinal)") 
print(df["Escolaridade"]) 
```

# Variaveis Binarias

São variaveis, com apenas dois valores possiveis. 

```python
print("Aprovado (variavel binária)") 
print(df['Aprovado'])
```

# variavel temporal.

representam tempo ou sequencia de eventos 

```python
print("Data_matricula (variavel temporal)") 
print(df['Data_matricula'])
```
