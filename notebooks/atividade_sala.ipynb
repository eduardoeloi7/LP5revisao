## Identificação e Remoção de Duplicatas

O código verifica se existem colunas com nomes repetidos e linhas duplicadas no dataset, removendo as duplicatas encontradas.

- `df.columns.duplicated()`: identifica nomes de colunas repetidos.
- `.tolist()`: transforma os nomes encontrados em uma lista.
- `if`: verifica se existem colunas duplicadas.
- `df.loc[:, ~df.columns.duplicated()]`: mantém apenas a primeira coluna de cada nome repetido.
- `df.duplicated().sum()`: conta as linhas duplicadas.
- `df.drop_duplicates()`: remove linhas repetidas, mantendo a primeira ocorrência.
- `df.shape`: informa a quantidade de linhas e colunas após a remoção.
- A última verificação confirma quantas linhas duplicadas restaram.

**Importância:** essa etapa evita redundâncias e ajuda a manter o dataset organizado para as análises posteriores.

```python
colunas_duplicadas = df.columns[df.columns.duplicated()].tolist()

print("Colunas duplicadas:", colunas_duplicadas)

if colunas_duplicadas:
    df = df.loc[:, ~df.columns.duplicated()]
    print("Colunas duplicadas removidas.")
else:
    print("Nenhuma coluna com nome duplicado foi encontrada.")

print("\nQuantidade de linhas duplicadas:")
print(df.duplicated().sum())

df = df.drop_duplicates()

print("\nDimensões após remover linhas duplicadas:")
print(df.shape)

print("\nQuantidade de linhas duplicadas restantes:")
print(df.duplicated().sum())
```

- `df.isna().sum()`: conta os valores ausentes em cada coluna.
- `ausentes[ausentes > 0]`: seleciona apenas as colunas com valores ausentes.
- `.sort_values(ascending=False)`: organiza os resultados do maior para o menor.
- `df.isna().mean() * 100`: calcula o percentual de valores ausentes por coluna.
- `.round(2)`: arredonda os percentuais para duas casas decimais.
- `df.dtypes[ausentes > 0]`: mostra os tipos de dados das colunas afetadas.
- `df.isna().sum().sum()`: calcula o total de valores ausentes em todo o dataset.

**Importância:** essa verificação permite avaliar a situação dos dados e identificar quais colunas ainda precisam de tratamento.

**Observação:** o código apenas analisa os valores ausentes, sem modificar o DataFrame.

```python
ausentes = df.isna().sum()

print("Quantidade de valores ausentes por coluna:")
display(ausentes[ausentes > 0].sort_values(ascending=False))

print("\nPercentual de valores ausentes por coluna:")
percentual = (df.isna().mean() * 100).round(2)
display(percentual[percentual > 0].sort_values(ascending=False))

print("\nTipos de dados das colunas com valores ausentes:")
display(df.dtypes[ausentes > 0])

print("\nTotal de valores ausentes no dataset:")
print(df.isna().sum().sum())
```

## Verificação e Remoção de Colunas Duplicadas

O código verifica se as colunas `cidade` e `cidade.1`, assim como `nota_final` e `nota_final.1`, possuem exatamente os mesmos valores.

- `.equals()`: verifica se duas colunas são idênticas, considerando os valores e suas posições.
- `if`: executa uma ação somente quando a condição é verdadeira.
- `df.drop(columns=[...])`: remove a coluna indicada, caso ela seja idêntica à outra.
- `df.columns.tolist()`: exibe a lista de colunas restantes no DataFrame.

Essa etapa evita manter colunas redundantes, deixando o conjunto de dados mais organizado.

**Importante:** se as colunas não forem totalmente idênticas, elas não serão removidas por esse código.

```python
print("cidade e cidade.1 são idênticas?")
print(df['cidade'].equals(df['cidade.1']))

print("\nnota_final e nota_final.1 são idênticas?")
print(df['nota_final'].equals(df['nota_final.1']))

if df['cidade'].equals(df['cidade.1']):
    df = df.drop(columns=['cidade.1'])

if df['nota_final'].equals(df['nota_final.1']):
    df = df.drop(columns=['nota_final.1'])

print("\nColunas restantes:")
print(df.columns.tolist())
```

## Identificação de Diferenças entre Colunas

O código identifica registros em que `cidade` e `cidade.1`, ou `nota_final` e `nota_final.1`, apresentam valores diferentes.

- `.ne()`: verifica quais valores são diferentes.
- `.isna()`: identifica valores ausentes.
- `&`: combina condições que precisam ser verdadeiras.
- `~`: inverte uma condição.
- `df.loc[]`: seleciona as linhas que atendem aos critérios.
- `id_aluno`: permite identificar o aluno relacionado à diferença encontrada.

A verificação desconsidera os casos em que ambas as colunas estão vazias. Assim, é possível identificar diferenças reais e avaliar se os dados precisam ser combinados antes de remover alguma coluna.

```python
print("Diferenças entre cidade e cidade.1:")
display(df.loc[
    df['cidade'].ne(df['cidade.1']) &
    ~(df['cidade'].isna() & df['cidade.1'].isna()),
    ['id_aluno', 'cidade', 'cidade.1']
])

print("\nDiferenças entre nota_final e nota_final.1:")
display(df.loc[
    df['nota_final'].ne(df['nota_final.1']) &
    ~(df['nota_final'].isna() & df['nota_final.1'].isna()),
    ['id_aluno', 'nota_final', 'nota_final.1']
])
```

## Tratamento de Colunas Complementares

O código verifica quantos valores estão preenchidos nas colunas `cidade.1` e `nota_final.1` e avalia se elas podem ser removidas sem perder informações.

- `.notna().sum()`: conta os valores preenchidos em cada coluna.
- `.isna().all()`: verifica se todos os valores de uma coluna estão ausentes.
- `.combine_first()`: combina duas colunas, priorizando os valores preenchidos da primeira e utilizando a segunda quando necessário.
- `if`: verifica se a condição é verdadeira antes de executar a remoção.
- `df.drop(columns=[...])`: remove a coluna indicada.
- `df.columns.tolist()`: mostra as colunas restantes após o tratamento.

Essa etapa ajuda a identificar colunas redundantes e evitar a perda de informações importantes durante a limpeza dos dados.

```python
print("Valores preenchidos em cidade.1:")
print(df['cidade.1'].notna().sum())

print("\nValores preenchidos em nota_final.1:")
print(df['nota_final.1'].notna().sum())

if df['cidade.1'].isna().all() or df['cidade'].equals(df['cidade.1'].combine_first(df['cidade'])):
    df = df.drop(columns=['cidade.1'])

if df['nota_final.1'].isna().all() or df['nota_final'].equals(df['nota_final.1'].combine_first(df['nota_final'])):
    df = df.drop(columns=['nota_final.1'])

print("\nColunas após o tratamento:")
print(df.columns.tolist())
```

## Tratamento da Coluna de Datas

O código converte a coluna `data_matricula` para o formato de data e cria duas novas colunas para facilitar a análise das matrículas por mês e ano.

- `pd.to_datetime()`: converte os valores para o formato de data.
- `errors='coerce'`: transforma valores inválidos em valores ausentes (`NaT`).
- `.dt.month`: extrai o mês da data.
- `.dt.year`: extrai o ano da data.
- `.head(10)`: exibe os primeiros 10 registros.
- `.dtype`: verifica o tipo de dado da coluna.
- `.isna().sum()`: contabiliza as datas ausentes após a conversão.

Essa etapa permite organizar as informações temporais e analisar a evolução das matrículas ao longo dos meses e anos.

```python
df['data_matricula'] = pd.to_datetime(df['data_matricula'], errors='coerce')

df['mes_matricula'] = df['data_matricula'].dt.month
df['ano_matricula'] = df['data_matricula'].dt.year

display(df[['data_matricula', 'mes_matricula', 'ano_matricula']].head(10))

print("\nTipo da coluna data_matricula:")
print(df['data_matricula'].dtype)

print("\nValores ausentes após a conversão:")
print(df['data_matricula'].isna().sum())
```

## Codificação One-Hot Encoding

O código transforma a coluna categórica `curso` em várias colunas numéricas, permitindo que essas informações sejam utilizadas em análises e modelos de Inteligência Artificial.

- `pd.get_dummies()`: converte categorias em colunas binárias.
- `columns=['curso']`: indica a coluna que será transformada.
- `dtype=int`: define os valores das novas colunas como números inteiros, geralmente `0` e `1`.
- `display(df.head())`: exibe os cinco primeiros registros após a transformação.
- `df.columns.tolist()`: mostra os nomes de todas as colunas após o One-Hot Encoding.

Cada curso passa a ter sua própria coluna: o valor `1` indica que o registro pertence àquela categoria, enquanto `0` indica que não pertence.

Essa técnica transforma dados categóricos em dados numéricos sem estabelecer uma ordem entre as categorias.

```python
df = pd.get_dummies(df, columns=['curso'], dtype=int)

display(df.head())

print("\nColunas após o One-Hot Encoding:")
print(df.columns.tolist())
```

## Codificação Label Encoding

O código substitui as categorias textuais da coluna `modalidade` por códigos numéricos, facilitando o uso dessas informações em análises de dados.

- `fillna('Não informado')`: substitui os valores ausentes por `"Não informado"`.
- `unique()`: identifica as categorias existentes na coluna.
- `sorted()`: organiza as categorias em ordem alfabética.
- `enumerate()`: atribui um número inteiro a cada categoria, começando em zero.
- `mapeamento`: cria um dicionário que relaciona cada categoria ao seu código numérico.
- `.map(mapeamento)`: aplica os códigos e cria a coluna `modalidade_encoded`.
- `display()`: exibe a modalidade original e seu código correspondente nos primeiros 10 registros.

Essa técnica transforma dados categóricos em números, mantendo a correspondência entre cada categoria e seu código.

**Observação:** os códigos representam categorias, não uma ordem de importância entre elas.

```python
df['modalidade'] = df['modalidade'].fillna('Não informado')

categorias = sorted(df['modalidade'].unique())

mapeamento = {categoria: codigo for codigo, categoria in enumerate(categorias)}

df['modalidade_encoded'] = df['modalidade'].map(mapeamento)

print("Mapeamento das categorias:")
print(mapeamento)

display(df[['modalidade', 'modalidade_encoded']].head(10))
```

## Comparação Gráfica Antes e Depois do Tratamento

O código cria gráficos para comparar o conjunto de dados original com o conjunto tratado, permitindo visualizar as mudanças após a limpeza dos dados.

### 1. Preparação dos dados

- `pd.read_csv()`: carrega o dataset original.
- `.copy()`: cria uma cópia para realizar os tratamentos sem modificar o DataFrame original.
- `.combine_first()`: combina colunas complementares, aproveitando os valores preenchidos.
- `.drop(columns=[...])`: remove as colunas duplicadas após a combinação.
- `.drop_duplicates()`: elimina registros repetidos.
- `.fillna()`: preenche valores ausentes com a mediana ou com `"Não informado"`.
- `pd.to_datetime()`: converte a coluna de matrículas para o formato de data.

### 2. Gráfico de distribuição das notas

O método `hist()` cria histogramas para comparar a distribuição das notas antes e depois do tratamento. O gráfico permite observar como o preenchimento de valores ausentes influencia a distribuição dos resultados.

### 3. Gráfico de quantidade de alunos por curso

- `value_counts()`: contabiliza os registros de cada curso.
- `pd.concat()`: reúne as contagens dos dois conjuntos de dados.
- `plot(kind='bar')`: cria um gráfico de barras para comparar as quantidades antes e depois do tratamento.

### 4. Gráfico de frequência versus nota

O método `scatter()` cria um gráfico de dispersão que relaciona a frequência dos alunos com suas notas finais. A comparação permite observar os registros disponíveis antes do tratamento e as alterações provocadas pelo preenchimento dos dados.

### 5. Gráfico de evolução das matrículas

- `dropna()`: seleciona registros que possuem datas válidas.
- `groupby()` e `size()`: agrupam as matrículas por mês e contam os registros de cada período.
- `pd.concat()`: reúne os resultados anteriores e posteriores ao tratamento.
- `plot(marker='o')`: cria um gráfico de linhas com marcadores para comparar a evolução mensal.

### 6. Ajustes finais dos gráficos

- `set_title()`: define o título de cada gráfico.
- `set_xlabel()` e `set_ylabel()`: identificam os eixos.
- `legend()`: exibe a legenda.
- `tick_params()`: ajusta a apresentação dos rótulos dos eixos.
- `plt.tight_layout()`: organiza os gráficos para evitar sobreposição.
- `plt.show()`: apresenta os quatro gráficos.

**Importância desta etapa:** a comparação visual permite avaliar como a remoção de duplicatas e o preenchimento de valores ausentes afetam a distribuição das notas, as quantidades por curso, a relação entre frequência e nota e a evolução das matrículas.

```python
df_antes = pd.read_csv('../dados/dataset_tratamento_dados.csv')
df_depois = df_antes.copy()

df_depois['cidade'] = df_depois['cidade'].combine_first(df_depois['cidade.1'])
df_depois['nota_final'] = df_depois['nota_final'].combine_first(df_depois['nota_final.1'])
df_depois = df_depois.drop(columns=['cidade.1', 'nota_final.1'])
df_depois = df_depois.drop_duplicates()

for coluna in ['idade', 'nota_final', 'frequencia_pct', 'satisfacao', 'horas_estudo_semana']:
    df_depois[coluna] = df_depois[coluna].fillna(df_depois[coluna].median())

for coluna in ['nome', 'cidade', 'curso', 'modalidade', 'ativo', 'entregou_projeto', 'dispositivo_principal']:
    df_depois[coluna] = df_depois[coluna].fillna('Não informado')

df_antes['data_matricula'] = pd.to_datetime(df_antes['data_matricula'], errors='coerce')
df_depois['data_matricula'] = pd.to_datetime(df_depois['data_matricula'], errors='coerce')

fig, ax = plt.subplots(2, 2, figsize=(16, 11))

ax[0, 0].hist(df_antes['nota_final'].dropna(), bins=10, alpha=0.6, label='Antes')
ax[0, 0].hist(df_depois['nota_final'].dropna(), bins=10, alpha=0.6, label='Depois')
ax[0, 0].set_title('Distribuição das notas')
ax[0, 0].set_xlabel('Nota final')
ax[0, 0].set_ylabel('Quantidade de alunos')
ax[0, 0].legend()

cursos = pd.concat([
    df_antes['curso'].fillna('Não informado').value_counts().rename('Antes'),
    df_depois['curso'].value_counts().rename('Depois')
], axis=1).fillna(0)

cursos.plot(kind='bar', ax=ax[0, 1])
ax[0, 1].set_title('Quantidade de alunos por curso')
ax[0, 1].set_xlabel('Curso')
ax[0, 1].set_ylabel('Quantidade')
ax[0, 1].tick_params(axis='x', rotation=45)

ax[1, 0].scatter(df_antes['frequencia_pct'], df_antes['nota_final'], alpha=0.6, label='Antes')
ax[1, 0].scatter(df_depois['frequencia_pct'], df_depois['nota_final'], alpha=0.6, label='Depois')
ax[1, 0].set_title('Frequência versus nota')
ax[1, 0].set_xlabel('Frequência (%)')
ax[1, 0].set_ylabel('Nota final')
ax[1, 0].legend()

matriculas_antes = df_antes.dropna(subset=['data_matricula']).groupby(
    df_antes['data_matricula'].dropna().dt.to_period('M')
).size()

matriculas_depois = df_depois.dropna(subset=['data_matricula']).groupby(
    df_depois['data_matricula'].dropna().dt.to_period('M')
).size()

matriculas = pd.concat([
    matriculas_antes.rename('Antes'),
    matriculas_depois.rename('Depois')
], axis=1).fillna(0).sort_index()

matriculas.index = matriculas.index.astype(str)
matriculas.plot(ax=ax[1, 1], marker='o')
ax[1, 1].set_title('Evolução das matrículas por mês')
ax[1, 1].set_xlabel('Mês/Ano')
ax[1, 1].set_ylabel('Quantidade de matrículas')
ax[1, 1].tick_params(axis='x', rotation=45)
ax[1, 1].legend()

plt.tight_layout()
plt.show()
```
